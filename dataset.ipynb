```python
from datasets import load_dataset
import pandas as pd
import numpy as np
```

```python

dataset = load_dataset("stanfordnlp/sst2")
train_df = dataset["train"].to_pandas()
test_df = dataset["validation"].to_pandas()

print(train_df)
print(test_df)
```

```python
train_df['char_length'] = train_df['sentence'].str.len()
train_df['word_count'] = train_df['sentence'].str.split().str.len()
test_df['char_length'] = test_df['sentence'].str.len()
test_df['word_count'] = test_df['sentence'].str.split().str.len()

print(train_df.head())
print(test_df.head())
```

```python
print(train_df.describe(include='all'))
print(test_df.describe(include='all'))
```
